### Make data decomposition

```python
fm = ForecastModel(7)
fm.prepare_data_decomposed()
```

### Train model for 7-day forecast horizon
- input length 126
- Load dataset from '1954-01-01' to '2008-12-31'
- Split data to train and validation dataset in '1986-09-01'

```python
fm = ForecastModel(7, context_length=126, tuned='default', split_date='1986-09-01', data_range=['1954-01-01', '2008-12-31'])
fm.train()
```

### Make prediction
- Select data range from 2023-01-01 to present 
- Save predictions to  (forecast_7)

```python
fm = ForecastModel(7, context_length=126, tuned='default', data_range=['2023-01-01',datetime.today().strftime('%Y-%m-%d')])
fm.todb(all=True)
```
